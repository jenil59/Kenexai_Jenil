```python
from  typing_extensions import TypedDict
from typing import Annotated

class State(TypedDict):
    graph_info:str
    
```

```python
def start_play(state:State):
    print("start play node is started :) ")
    
    return {'graph_info':state['graph_info']+"I am Playing now."}

def cricket(state:State):
    print("cricket game is started :) ")
    
    return {'graph_info':state['graph_info']+"Cricket "}

def football(state:State):
    print("football game is started :) ")
    
    return {'graph_info':state['graph_info']+" football."}
```

```python
import random
from typing import Literal

def random_play(state:State)->Literal['cricket','football']:
    graph_info  = state['graph_info']
    
    if random.random() > 0.5:
        return 'cricket'
    else:
        return 'football'
      
```

```python
from langgraph.graph import StateGraph,START,END
```

```python
graph = StateGraph(State)

graph.add_node("start_play",start_play)
graph.add_node("cricket",cricket)
graph.add_node("football",football)


graph.add_edge(START,"start_play")
graph.add_conditional_edges("start_play",random_play)
graph.add_edge("cricket",END)
graph.add_edge("football",END)


graph_builder = graph.compile() 

```

```python
graph_builder.invoke({'graph_info':"Hello ,Good Afternnon."})
```
